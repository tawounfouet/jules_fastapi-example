# FastAPI Feature Based API Test

This notebook demonstrates how to interact with the API endpoints.

```python
import requests
import json

BASE_URL = "http://localhost:8000/api/v1"
```

## 1. Create User

```python
email = "notebook@example.com"
password = "password123"
user_data = {"email": email, "password": password}

response = requests.post(f"{BASE_URL}/users/", json=user_data)
print(response.status_code)
print(response.json())
```

## 2. Login & Get Token

```python
login_data = {"username": email, "password": password}
response = requests.post(f"{BASE_URL}/auth/token", data=login_data)
print(response.status_code)
token_data = response.json()
print(token_data)
access_token = token_data.get("access_token")
headers = {"Authorization": f"Bearer {access_token}"}
```

## 3. Upload Avatar (Requires Minio/S3)

```python
# Create a dummy file
with open("avatar.txt", "w") as f:
    f.write("dummy image content")

files = {'file': ('avatar.txt', open('avatar.txt', 'rb'), 'text/plain')}
response = requests.post(f"{BASE_URL}/users/me/avatar", headers=headers, files=files)
print(response.status_code)
print(response.json())
```

## 4. Create Blog Post

```python
post_data = {"title": "Notebook Post", "content": "Created from Jupyter"}
response = requests.post(f"{BASE_URL}/posts/", json=post_data, headers=headers)
print(response.status_code)
post = response.json()
print(post)
post_id = post.get("id")
```

## 5. Add Comment

```python
comment_data = {"content": "Nice post!", "post_id": post_id}
response = requests.post(f"{BASE_URL}/comments/", json=comment_data, headers=headers)
print(response.status_code)
print(response.json())
```

## 6. Logout

```python
response = requests.post(f"{BASE_URL}/auth/logout", headers=headers)
print(response.status_code)
print(response.json())
```

## 7. Verify Logout (Should fail)

```python
response = requests.get(f"{BASE_URL}/users/", headers=headers)
print(response.status_code) # Should be 401
print(response.json())
```
